# Create a dataset for EVASPA with Sentinel2 and ECOSTRESS data 

### Imports

```python
%load_ext autoreload
%autoreload 2
```

```python
import os
```

```python
from etdataset.api import create_dataset
from etdataset.plot import plot_images
from etdataset.writer import export_matlab, write_dataset
```

```python
if os.environ["ETDATASET_TEST_DATA_PATH"] is None:
    raise Exception("Variable ETDATASET_TEST_DATA_PATH must be set")
etdataset_data_path = os.environ["ETDATASET_TEST_DATA_PATH"]
```

### Inputs 
* Sentinel2 product path
* ECOSTRESS product path

```python
# Product paths
s2_path = os.path.join(
    etdataset_data_path,
    "Sentinel2",
    "SENTINEL2B_20230402-114740-682_L2A_T28PCA_C_V3-1",
)
eco_path = os.path.join(
    etdataset_data_path,
    "Ecostress",
    "ECOv002_L2T_LSTE_26870_001_28PCA_20230402T115902_0710_01",
)
era5_path = os.path.join(
    etdataset_data_path,
    "era5",
    "download_era5_2023-04-02_14.57_-16.96_13.37_-15.73.nc",
)
```

```python
# Directory path for results
output_dir = "out"
os.makedirs(output_dir, exist_ok=True)
```

### Create dataset

```python
# To take or not into account mask, use option use_mask
ds = create_dataset(vis_path=s2_path, tir_path=eco_path, radiation=era5_path)
```

```python
plot_images(ds)
```

### Write dataset

```python
# Tif format
write_dataset(ds, directory=output_dir)
```

```python
# Matlab
export_matlab(ds, directory=output_dir)
```

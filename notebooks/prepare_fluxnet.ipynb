## Fluxnet stations

Load the list of valid Fluxnet stations and their metadata from the CSV file.

```python
valid_stations_list = get_fluxnet_stations_list()
```

Get the stations list

```python
valid_stations_list
```

### Filering

 Check the available stations by source code (e.g "AmeriFlux", "ICOS", "TERN")

For AmeriFlux

```python
ameriflux_station = filter_fluxnet_stations_by_sources("AmeriFlux")
```

For ICOS

```python
icos_station = filter_fluxnet_stations_by_sources("ICOS")
```

Filter and get Fluxnet stations that have available data between 2023 and 2024

```python
stations_by_year = filter_fluxnet_stations_by_years(2023, 2024)
```

Filter the Ameriflux stations to keep only the selected group available for 2023-2024

```python
ameriflux_2023_2024 = ameriflux_station[
    ameriflux_station.index.isin(stations_by_year.index)
]
```

Get the stations' location from the stations list (and you can save it as GeoPackage)

```python
gdf = get_fluxnet_station_location(ameriflux_2023_2024["id"])
```

Save the GeoDataFrame in a GeoPackage

```python
create_geopckg_from_gdf(gdf)
```

### Plot

Plot an interactive map of the stations from the GeodataFrame created

```python
plot_station_map(gdf)
```

### Download

Download station datas ("all" or list of stations)

```python
download_fluxnet_data(stations="CA-ARF", output_dir="fluxnet_data")
```

Get the station configuration (name, source, latitude, longitude)

```python
cfg = get_fluxnet_stations_config("CA-ARF")
```

Get the correct CSV file path for the given station ID from the data directory

```python
path = get_fluxnet_archive(cfg.id, data_dir="fluxnet_data")
```

Get datas from csv

```python
data = read_fluxnet_data(path)
```

### Save

Save the station's datas in a csv

```python
save_fluxnet_station(cfg, data, "fluxnet_processed")
```

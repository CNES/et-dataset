```python
import logging
import os
```

```python
from etdataset.api import download, download_aux, search
from etdataset.logging import LoggerManager
from etdataset.provider import Collection
from etdataset.utils import read_product_list
```

```python
LoggerManager.set_level(logging.DEBUG)
```

### Donwload directory

```python
output_dir = os.path.join(os.getcwd(), "out")
os.makedirs(output_dir, exist_ok=True)
```

### Product list

```python
products = read_product_list("products.csv", 4326)
```

```python
products
```

```python
download(products, output_dir=output_dir)
```

### Search and download Sentinel2 products

```python
res = search(Collection.SENTINEL2, "2023-03-28", "2023-03-28", tile_id="28PCA")
```

```python
res
```

```python
download(res, output_dir=output_dir)
```

### Search and download ECOSTRESS products

```python
res = search(Collection.ECOSTRESS, "2023-02-13", "2023-03-01", tile_id="28PCA")
```

```python
res
```

```python
download(res, output_dir=output_dir)
```

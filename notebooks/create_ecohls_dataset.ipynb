# Create a dataset for EVASPA with HLS and ECOSTRESS data 

### Imports

```python
%load_ext autoreload
%autoreload 2
```

```python
import os
```

```python
from etdataset.api import create_dataset
from etdataset.plot import plot_images
from etdataset.writer import write_dataset, export_matlab
```

### Inputs 
* HLS product path
* ECOSTRESS product path

```python
# Product paths
data_path = os.environ["DATA_PATH"]
hlsl_path = os.path.join(
    data_path, "HLSL","HLS.L30.T28PCA.2023086T112728.v2.0"
)
hlss_path = os.path.join(
    data_path, "HLSS","HLS.S30.T28PCA.2023092T113319.v2.0"
)
eco_path = os.path.join(
    data_path, "ECOSTRESS", "ECOv002_L2T_LSTE_26870_001_28PCA_20230402T115902_0710_01"
)
```

```python
# Directory path for results
output_dir = "out"
os.makedirs(output_dir, exist_ok=True)
```

### Create HLSL/ECOSTRESS dataset

```python
from etdataset.reader import HLSReader
ds = HLSReader(hlsl_path)
```

```python
# To take or not into account mask, use option use_mask
ds = create_dataset(vis_path=hlsl_path, tir_path=eco_path, use_mask=True)
```

```python
plot_images(ds)
```

```python
# Tif format
write_dataset(ds, directory=output_dir)
```

```python
# Matlab
export_matlab(ds, directory=output_dir)
```

### Create HLSS/ECOSTRESS dataset

```python
# To take or not into account mask, use option use_mask
ds = create_dataset(vis_path=hlss_path, tir_path=eco_path, use_mask=True)
```

```python
plot_images(ds)
```

```python
# Tif format
write_dataset(ds, directory=output_dir)
```

```python
# Matlab
export_matlab(ds, directory=output_dir)
```

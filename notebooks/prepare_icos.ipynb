## ICOS stations

Get a csv file with all the valid ICOS stations and their metadata

```python
get_csv_with_valid_icos_stations()
```

### Vizualize available station

Get the stations list

```python
get_station_list()
```

Check the available stations by country number

```python
filter_stations_by_country_code("FR")
```

### Download

Download icos station datas ("all" or list of stations)

```python
download_icos_station(["DE-Geb", "CH-Dav"])  # "all"
```

Get the station confiugration (name, country, latitude, longitude and elevation of the station)

```python
cfg = get_stations_config("DE-Geb")
```

Get datas from csv

```python
data = read_csv_data(cfg)
```

Filter only valid datas (Air temperature Ta and relative humidity Rh) of the station

```python
data = filter_valid_data(data)
```

### Compute auxiliary temperature

Compute the dewpoint temperature (Td) with Ta and Rh

```python
td = compute_dewpoint_temp(data["TA"], data["RH"])
```

Add dewpoint temperature

```python
data["TD"] = td
```

### Save

Save the station's datas in a csv

```python
csv_path = save_station_data(cfg, data)
```

Create a list with chosen stations

```python
stations_list = [
    "BE-Bra",
    "BE-Vie",
    "CH-Dav",
    "CZ-BK1",
    "DE-Brs",
    "DE-Geb",
    "DE-GsB",
    "DE-Msr",
    "DE-RuW",
    "FI-Tvm",
    "FR-Aur",
    "FR-CLt",
]
```

Get the stations' location from the stations list (and you can save it as GeoPackage)

```python
gdf = get_station_location(stations_list)
```

Save the GeoDataFrame in a GeoPackage

```python
create_geopckg_from_gdf(gdf)
```

### Plot

Plot an interactive map of the stations from the GeodataFrame created

```python
plot_station_map(gdf)
```

Plot the Icos air temperature and the dew point temperature times series
